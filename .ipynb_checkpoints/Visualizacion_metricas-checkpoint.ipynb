```python
import json
import networkx as nx
import matplotlib.pyplot as plt

# Cargar JSON
with open("user_3_hotel_45.json", "r", encoding="utf-8") as f:
    data = json.load(f)

G = nx.Graph()

# Añadir nodos solo si tienen ID válido
for n in data["nodes"]:
    nid = n.get("id")
    if nid is not None:     # filtrar nodos inválidos
        G.add_node(str(nid))

# Añadir relaciones
for r in data["relationships"]:
    s = r.get("start_node")
    t = r.get("end_node")
    if s is not None and t is not None:
        G.add_edge(str(s), str(t))

# Layout rápido
pos = nx.random_layout(G)

plt.figure(figsize=(15, 15))
nx.draw(G, pos, node_size=5, edge_color="gray", alpha=0.5)
plt.title("Grafo completo (visualización rápida)")
plt.show()
```

```python
import json
import networkx as nx
import matplotlib.pyplot as plt

# Cargar JSON
with open("user_3_hotel_45.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# Filtrar nodos con id válido
valid_nodes = [n for n in data["nodes"] if n.get("id") is not None]

# Seleccionar 500 nodos válidos
selected_nodes = { str(n["id"]) for n in valid_nodes[:500] }

# Crear subgrafo
G = nx.Graph()

for nid in selected_nodes:
    G.add_node(nid)

# Añadir relaciones entre nodos seleccionados
for r in data["relationships"]:
    s = r.get("start_node")
    t = r.get("end_node")
    if s in selected_nodes and t in selected_nodes:
        G.add_edge(str(s), str(t))

# Layout físico
pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(15, 15))
nx.draw(G, pos, node_size=20, edge_color="gray", alpha=0.7)
plt.title("Subgrafo de 500 nodos")
plt.show()
```
